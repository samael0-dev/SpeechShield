# 01 - Social Media Toxicity Data Exploration
Explore class balance, platform distribution, and text length characteristics.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from scripts.scrape_data import build_starter_dataset
from src.model.utils import TOXICITY_CATEGORIES

raw_path = PROJECT_ROOT / "data" / "raw" / "sample_social_comments.csv"
df = pd.read_csv(raw_path) if raw_path.exists() else build_starter_dataset()
df.head()
```

```python
print(f"Total rows: {len(df)}")
print("\nPlatform distribution:")
print(df["platform"].value_counts())

label_counts = df[list(TOXICITY_CATEGORIES)].sum()
print("\nCategory counts:")
print(label_counts)

label_counts.plot(kind="bar", title="Toxicity Category Counts", figsize=(8, 4))
plt.ylabel("Samples")
plt.tight_layout()
plt.show()
```
