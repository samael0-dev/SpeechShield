# 02 - Social Media Preprocessing Experiments
Evaluate hashtag unpacking, emoji demoji translation, contraction expansion, and slur de-obfuscation.

```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_pipeline.cleaner import SocialTextCleaner

cleaner = SocialTextCleaner(use_ekphrasis=True)
samples = [
    "You're such a clown 🤡 #IdentityHate https://example.com @user123",
    "Shut up you @ss $lut stfu gtfo",
    "GG everyone! That match was awesome 🙌🔥 #ProGaming",
    "Whyyyyy are people so b!tch online smh"
]

rows = []
for text in samples:
    res = cleaner.clean(text)
    rows.append({
        "raw_text": res.raw_text,
        "cleaned_text": res.cleaned_text,
        "hashtags": res.hashtags,
        "emojis": res.emojis,
        "obfuscation_hits": res.obfuscation_hits,
        "sentiment_score": res.sentiment_score,
    })

pd.DataFrame(rows)
```
